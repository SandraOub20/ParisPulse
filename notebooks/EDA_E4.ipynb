## Building cases.csv

```python
# protocol config
PUBLICATION_LAG_HOURS = 24  
BUDGETS = [24, 72, 168]  
PERIODS = {
    "validation": ("2026-06-01", "2026-09-30"),
    "test": ("2026-10-01", "2026-11-14"),   
}
ROLES = {
    "S_train": ("known_site", ["validation", "test"]),
    "S_dev_new": ("new_site", ["validation"]),
    "S_test_new": ("new_site", ["test"]),
}
COLUMNS = [
    "case_id", "counter_id", "site_group_id", "origin", "scenario",
    "temporal_split", "spatial_role", "local_budget_hours", "support_start",
    "support_end", "protocol_id", "snapshot_id",
]
```

```python
selected = quality[quality["period_role"] == "selection"]           
selected = selected[selected["selection_status"] != "excluded"]     
print(len(selected), "counters")
print(selected["site_role"].value_counts())
```

```python
def get_origins(start, end):
    """One midnight (Paris time) per day, converted to UTC"""
    days = pd.date_range(start, end, freq="D")
    return days.tz_localize(TIMEZONE).tz_convert("UTC")


def to_text(ts):
    return ts.strftime("%Y-%m-%dT%H:%M:%S+00:00")


def make_case_id(counter_id, origin, scenario, budget): 
    b = "" if budget is None else f"_b{budget}"
    return f"{counter_id}_{origin:%Y%m%dT%H}_{scenario}{b}"
```

```python
rows = []
for _, c in selected.iterrows():
    scenario, splits = ROLES[c["site_role"]]
    for split in splits:
        for origin in get_origins(*PERIODS[split]):
            budgets = BUDGETS if scenario == "new_site" else [None]
            for budget in budgets:
                if budget is None:
                    support_start, support_end = "", ""
                else:
                    end = origin - pd.Timedelta(hours=PUBLICATION_LAG_HOURS)
                    start = end - pd.Timedelta(hours=budget)
                    support_start, support_end = to_text(start), to_text(end)
                rows.append({
                    "case_id": make_case_id(c["counter_id"], origin, scenario, budget),
                    "counter_id": c["counter_id"],
                    "site_group_id": c["site_group_id"],
                    "origin": to_text(origin),
                    "scenario": scenario,
                    "temporal_split": split,
                    "spatial_role": c["site_role"],
                    "local_budget_hours": "" if budget is None else budget,
                    "support_start": support_start,
                    "support_end": support_end,
                    "protocol_id": PROTOCOL_ID,
                    "snapshot_id": SNAPSHOT_ID,
                })
cases = pd.DataFrame(rows, columns=COLUMNS)
print(len(cases), "cases")
```

checks on cases.csv

```python
print("duplicated case_id:", cases["case_id"].duplicated().sum())
print("site groups with several roles:", (cases.groupby("site_group_id")["spatial_role"].nunique() > 1).sum())

# no train counter inside a reserved group (leakage)
reserved = set(site_groups["dev_new"]) | set(site_groups["test_new"])
train_cases = cases[cases["spatial_role"] == "S_train"]
print("train cases in a reserved group:", train_cases["site_group_id"].isin(reserved).sum())

# all origins must have complete 24h coverage in counts.csv
last_data = pd.to_datetime(counts["timestamp"], utc=True).max()
last_target = pd.to_datetime(cases["origin"], utc=True) + pd.Timedelta(hours=23)
print("cases missing data at the end:", (last_target > last_data).sum())

print(cases.groupby(["scenario", "temporal_split", "local_budget_hours"], dropna=False).size())
```
